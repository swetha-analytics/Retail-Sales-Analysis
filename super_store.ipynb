```python
import pandas as pd

df = pd.read_csv("Sample - Superstore.csv")
```

```python
#Which region has the highest Average Order Value (AOV)?
region_aov =( df.groupby("Region")
    .agg(
        Total_sales = ("Sales","sum"),
        unique_orders = ("Order ID", "nunique")
    )
)
region_aov["AOV"] = (region_aov["Total_sales"]/region_aov["unique_orders"])
region_aov = region_aov.sort_values("AOV", ascending=False)

print (region_aov)
```

```python
df["Order Date"] = pd.to_datetime(
    df["Order Date"],
    format="%d-%m-%Y"
)

df["Year"] = df["Order Date"].dt.year
df["Month"] = df["Order Date"].dt.month
```

```python
#Which months had the highest and lowest total Sales and Profit?
monthly_performance = (
    df.groupby(["Year", "Month"])
      .agg(
          total_sales=("Sales", "sum"),
          total_profit=("Profit", "sum")
      )
)

monthly_performance = monthly_performance.sort_index()

print(monthly_performance)
```

```python
#At what discount level does profit turn negative for each category?
discount_analysis = (
    df.groupby(["Category", "Discount"])
      .agg(
          total_profit=("Profit", "sum"),
          total_sales=("Sales", "sum"),
          average_profit=("Profit", "mean")
      )
)
discount_analysis[discount_analysis["total_profit"] < 0]
```

```python
#Which Region and Ship Mode combinations have the longest average delivery time?

df["Order Date"] = pd.to_datetime(
    df["Order Date"],
    format="%d-%m-%Y")
df["Ship Date"] = pd.to_datetime(
    df["Ship Date"],format="%d-%m-%Y")
df["Delivery Days"] = (
    df["Ship Date"] - df["Order Date"]
).dt.days

shipping_analysis = (
    df.groupby(["Region", "Ship Mode"])
      .agg(
          avg_delivery_days=("Delivery Days", "mean"),
          max_delivery_days=("Delivery Days", "max"),
          total_orders=("Order ID", "nunique")
      )
      .reset_index()
      .sort_values(
          "avg_delivery_days",
          ascending=False
      )
)

print(shipping_analysis)
```

```python
#Based on the historical monthly sales trend, what will the estimated total sales be for the next 3 months?

monthly_sales =( 
    df.groupby(["Year","Month"])
    .agg(
        Total_sales = ("Sales","sum")
        )           
.reset_index()
.sort_values(["Year","Month"])
)              
print(monthly_sales)

monthly_sales["Month_num"] = range(len(monthly_sales))
```

```python
from sklearn.linear_model import LinearRegression
```

```python
monthly_sales["Month_Number"] = range(len(monthly_sales))
```

```python
X = monthly_sales[["Month_Number"]]
y = monthly_sales["Total_sales"]

model = LinearRegression()
model.fit(X, y)
```

```python
future_months = pd.DataFrame({
    "Month_Number": [
        len(monthly_sales),
        len(monthly_sales) + 1,
        len(monthly_sales) + 2
    ]
})

forecast = model.predict(future_months)

print(forecast)

```

```python
#### Key Insight

A baseline Linear Regression model was used to estimate sales for the next three months based on historical monthly sales trends.

The forecast predicts a gradual increase in sales. However, this model should be considered a baseline because it does not account for seasonality, promotions, or other external business factors.
```
